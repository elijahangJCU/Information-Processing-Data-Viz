## Student performance Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load and read data
sp = pd.read_csv("Performance_of_Stuednts.csv")
print(sp.head())
print(sp.info())
```

#### Some Light EDA

```python
gender = sp['gender'].value_counts().reset_index()
print(gender)

male_percent = (gender.loc[1, 'count'] / gender.loc[1, 'count'] + gender.loc[0, 'count']) * 100
female_percent = (gender.loc[0, 'count'] / gender.loc[0, 'count'] + gender.loc[1, 'count']) * 100

print(f"Male: {male_percent:.2f}%")
print(f"Female: {female_percent:.2f}%")
```

```python
fig, ax = plt.subplots()

ax.pie([male_percent, female_percent], labels=['Male', 'Female'], autopct='%1.1f%%')
ax.set_title("Proportion of Male and Female Students")
plt.show()
```

```python
parental_education = sp['parental education level'].value_counts().reset_index()
print(parental_education)
```

```python
fig, ax = plt.subplots()

# Pass the 'count' column.
# 'autopct' handles the math so you don't have to!
ax.pie(parental_education['count'],
       labels=parental_education['parental education level'],
       autopct='%1.1f%%',
       startangle=140) # Rotates the chart for better readability

ax.set_title('Proportion of Parental Education Levels')
plt.show()
```

```python
lunch = sp['lunch'].value_counts().reset_index()
print(lunch)
```

```python
fig, ax = plt.subplots()

ax.pie(lunch['count'], labels=lunch['lunch'], autopct='%1.1f%%', startangle=140)
ax.set_title('Proportion of Lunch Types')
plt.show()
```

```python
test_prep_course = sp['test prep course'].value_counts().reset_index()
print(test_prep_course)
```

```python
fig, ax = plt.subplots()

ax.pie(test_prep_course['count'], labels= test_prep_course['test prep course'], autopct='%1.1f%%', startangle=140)
ax.set_title('Proportion of Test Prep Courses')
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 5))

sns.histplot(sp['math score'], bins=100, kde=True, ax=ax[0])
ax[0].axvline(sp['math score'].median(), linestyle='dashed', color='red', label=f"Median {sp['math score'].median()}")
ax[0].set_title('Math Score Distribution')
ax[0].legend()

sns.histplot(sp['science score'], bins=100, kde=True, ax=ax[1])
ax[1].axvline(sp['science score'].median(), linestyle='dashed', color='red', label=f"Median {sp['science score'].median()}")
ax[1].set_title('Science Score Distribution')
ax[1].legend()

sns.histplot(sp['english score'], bins=100, kde=True, ax=ax[2])
ax[2].axvline(sp['english score'].median(), linestyle='dashed', color='red', label=f"Median {sp['english score'].median()}")
ax[2].set_title('English Score Distribution')
ax[2].legend()

plt.show()
```
