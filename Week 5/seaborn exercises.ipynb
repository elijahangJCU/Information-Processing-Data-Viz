### Exercise 1

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales = {
    'Year': [2018, 2019, 2020, 2021],
    'Sales': [100, 120, 150, 170],
    'Expenses': [80, 90, 110, 130],
    'Region': ['East', 'East', 'West', 'West']
}

sales_data = pd.DataFrame(sales)
```

```python
%matplotlib inline
sales_data.plot(x='Year', y='Sales', kind='bar')
sales_data.plot(x='Year', y='Expenses', kind='line')
```

### Exercise 2

```python
%matplotlib inline
fig, ax = plt.subplots()
ax.bar(sales_data['Year'], sales_data['Sales'], label='Sales')
ax.set_xlabel('Years')
ax.set_ylabel('Sales')
ax.set_xticks(sales_data['Year'])
ax.set_title('Sales vs Year')
ax.legend()
```

```python
%matplotlib inline

fig, ax = plt.subplots()
ax.bar(sales_data['Year'], sales_data['Sales'], color='blue', label='Sales')
ax.bar(sales_data['Year'], sales_data['Expenses'], bottom=sales_data['Sales'], color='red', label='Expenses')

ax.set_xlabel('Years')
ax.set_ylabel('Sales and Expenses')
ax.set_xticks(sales_data['Year'])
ax.legend()
```

### Exercise 3

```python
%matplotlib inline

fig, ax = plt.subplots()

ax.hist(sales_data['Sales'], bins=10, label='Sales', edgecolor='pink')
ax.set_xlabel('Sales')
ax.set_ylabel('Count')
ax.set_title('Sales distribution')
```

### Exercise 4

```python
student_data = pd.DataFrame({
    'Hours': [2,4,6,8,5],
    'Score': [50,65,80,90,70],
    'Gender': ['M','F','F','M','F'],
    'Study_Type': ['Group','Solo','Solo','Group','Group']
})
```

```python
%matplotlib notebook

fig, ax = plt.subplots()

sns.scatterplot(x='Hours', y='Score', hue='Gender', style='Study_Type', data=student_data, size='Hours', ax=ax)
ax.set_title('Student study hours vs score')
ax.set_xlabel('Hours')
ax.set_ylabel('Score')
ax.legend()

plt.show()
```

### Exercise 5

```python
student_data_expanded = pd.DataFrame({
    'Hours': [2,2,4,4,6,6,8,8,5,5],
    'Score': [50,55,65,60,80,78,90,85,70,72]
})
```

```python
fig, ax = plt.subplots()

sns.lineplot(x='Hours', y='Score', data=student_data_expanded, ax=ax)
ax.set_title('Score trend')
ax.set_xlabel('Hours')
ax.set_ylabel('Score')

plt.show()
```

### Exercise 7

```python
fig1, ax1 = plt.subplots()

sns.countplot(x='Gender', hue='Gender', data=student_data, ax=ax1)
ax1.set_xlabel('Gender')
ax1.set_ylabel('Count')
ax1.set_title('Gender vs Gender')

fig2, ax2 = plt.subplots()

sns.countplot(x='Study_Type', data=student_data, hue='Gender', ax=ax2)
ax2.set_xlabel('Study Type')
ax2.set_ylabel('Count')
ax2.set_title('Study Type vs Study Type')

plt.show()
```

### Exercise 8

```python
# box plot

fig1, ax1 = plt.subplots()

sns.boxplot(x='Study_Type', y='Score', data=student_data, ax=ax1)
ax1.set_xlabel('Study Type')
ax1.set_ylabel('Score')
ax1.set_title('Study Type vs Score')

fig2, ax2 = plt.subplots()

sns.violinplot(x='Study_Type', y='Score', data=student_data, ax=ax2)
ax2.set_xlabel('Study Type')
ax2.set_ylabel('Score')
ax2.set_title('Study Type vs Score')

plt.show()
```

### Exercise 9

```python
fig, ax = plt.subplots()

sns.histplot(x='Score', hue='Gender', data=student_data, ax=ax, kde=True)
ax.set_xlabel('Score')
ax.set_ylabel('Count')
ax.set_title('Score distribution')

plt.show()
```

### Exercise 10

```python
fig, ax = plt.subplots()

sns.kdeplot(x='Score', hue='Gender', data=student_data, ax=ax)
ax.set_xlabel('Score')
ax.set_ylabel('Count')
ax.set_title('Score distribution')
plt.show()
```

### Exercise 11

```python
corr_matrix = student_data.corr(numeric_only=True)

fig, ax = plt.subplots()

sns.heatmap(corr_matrix, cmap='coolwarm', annot=True, ax=ax)
ax.set_title('Hours and study correlation')

plt.show()
```

### Exercise 12

```python
corr_matrix = student_data.corr(numeric_only=True)

cg = sns.clustermap(
    corr_matrix,
    cmap='coolwarm',
    annot=True
)

cg.fig.suptitle('Cluster Map of Numeric Variables')

plt.show()
```

### Exercise 13

```python
fig, ax = plt.subplots()

sns.regplot(x='Hours', y='Score', data=student_data, ax=ax, color='red')

ax.set_title('Hours and score correlation')

plt.show()
```

### Exercise 14

```python
g = sns.lmplot(x='Hours', y='Score', data=student_data, hue='Gender', col='Study_Type')

g.fig.suptitle('Regression of score on hours by study type')
g.fig.subplots_adjust(top=0.85)

plt.show()
```
